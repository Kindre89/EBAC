### 1. Explorando a base de gorjetas

a. Na aula fizemos uma regressão de ```tip``` explicada por ```net_bill```. Carregue os pacotes e a base, crie as variáveis necessárias como na aula, e reproduza esta regressão.

b. Reproduza também o gráfico de dispersão dessas duas variáveis.

````python
# Import the necessary modules and libraries
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from seaborn import load_dataset
import pandas as pd

import statsmodels.formula.api as smf

````

````python
#! pip install statsmodels
````

````python
tips = sns.load_dataset("tips")
tips['tip_pct'] = tips['tip'] / (tips['total_bill'] - tips['tip'])
tips["net_bill"] = (tips["total_bill"] - tips["tip"])
tips.head()
````

````python
sns.regplot(y="tip", x="net_bill", data=tips[tips["tip_pct"]<.5]) 

x = np.linspace(0,tips["net_bill"].max(),100)
y = .1*x

plt.plot(x, y, "-r", label="linha dos 10%")

plt.xlim(left = 0)
plt.ylim(bottom = 0)
````

````python
reg = smf.ols("tip ~ net_bill", data = tips).fit()
reg.summary()
````

````python
dados_novos = tips[0:5]
dados_novos
````

````python
reg.predict(dados_novos)
````

````python
type(reg)
````

````python
reg.params
````

### 2. Mudança de perspectiva

Agora ajuste um modelo de regressão de ```tip_pct``` por ```net_bill```. Construa o gráfico de dispersão correspondente.

````python
reg_2 = smf.ols("tip_pct ~ net_bill", data = tips).fit()
reg_2.summary()
````

````python
sns.regplot(y="tip_pct", x="net_bill", data=tips[tips["tip_pct"]<.5]) 

x = np.linspace(0,tips["net_bill"].max(),100)
y = .1*x

plt.plot(x, y, "-r", label="linha dos 10%")

plt.xlim(left = 0)
plt.ylim(bottom = 0)
````

### 3. Comente

Compare os dois gráficos e comente (de forma objetiva e suscinta) a relação entre eles e a diferença do ajuste. Podemos comparar diretamente os $R^2$s?

````python
print(f"{reg.rsquared}")
print(f"{reg_2.rsquared}")
````

Conforme observado na célula acima, o erro quadrado do primeiro modelo é significativamente maior que o do segundo, no qual utilizamos a coluna tip_pct. Isso indica que o segundo modelo está melhor ajustado aos nossos dados.
